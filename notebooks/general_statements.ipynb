```python
from pathlib import Path

import polars as pl
from IPython.display import display

dir_path = Path("../docs").joinpath("accounts").joinpath("general")

dir_path.mkdir(exist_ok=True, parents=True)
```

```python
bs_file_path = dir_path.joinpath("balance_statement_accounts.csv")

df_bs = pl.read_parquet("../.data/stagings/general_balance_statement/*.parquet").filter(
    pl.col("value") != 0
)

if not bs_file_path.exists():
    total_variants = df_bs["variant"].n_unique()

    df_account = df_bs.group_by("account").agg(
        pl.when(pl.col("variant").n_unique() == total_variants)
        .then(pl.lit("all_variants"))
        .otherwise(pl.col("variant").first())
        .alias("variant")
    )
    df_account.write_csv(bs_file_path)
    display(df_account.n_unique())
```

```python
is_file_path = dir_path.joinpath("income_statement_accounts.csv")

df_is = pl.read_parquet("../.data/stagings/general_income_statement/*.parquet").filter(
    pl.col("value") != 0
)

if not is_file_path.exists():
    total_variants = df_is["variant"].n_unique()

    is_account = df_is.group_by("account").agg(
        pl.when(pl.col("variant").n_unique() == total_variants)
        .then(pl.lit("all_variants"))
        .otherwise(pl.col("variant").first())
        .alias("variant")
    )
    is_account.write_csv(is_file_path)
    display(is_account.n_unique())
```

```python
import json
from pathlib import Path

import polars as pl
```

```python
def generate_accounts_schema(df: pl.DataFrame, path: Path) -> None:
    account_names = (
        df.select("account").drop_nulls().unique().get_column("account").to_list()
    )

    schema = {
        "$schema": "https://json-schema.org/draft-07/schema#",
        "type": "string",
        "enum": account_names,
    }

    path.write_text(
        json.dumps(schema, ensure_ascii=False, indent=2),
        encoding="utf-8",
    )

    print(f"Schema written to {path.resolve()}")
```

```python
generate_accounts_schema(
    pl.read_parquet(
        "../.data/transformations/enriched_accounts/general/balance_statement/current_and_non_current/accounts.parquet"
    ),
    Path(
        "../src/idx_hamilton/schema/general/balance_statement/current_and_non_current/account.schema.json"
    ),
)
```
